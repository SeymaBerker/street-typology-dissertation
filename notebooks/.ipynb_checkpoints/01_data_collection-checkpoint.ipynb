```python
import osmnx as ox
import geopandas as gpd
import matplotlib.pyplot as plt

print("Success.")
```

```python
# Download the street network for City of London
G = ox.graph_from_place("City of London, UK", network_type="drive")

# Convert to a table
nodes, edges = ox.graph_to_gdfs(G)

print(f"Done. Number of street segments: {len(edges)}")
edges.head()
```

```python
# Save your streets so you never need to download again
edges.to_file("/home/jovyan/work/data/london_streets.gpkg", driver="GPKG")
print("Streets saved to data folder.")
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

edges.plot(ax=ax, linewidth=0.8, color="#378ADD", alpha=0.8)

ax.set_title("City of London — Street Network", fontsize=14)
ax.set_axis_off()

plt.savefig("/home/jovyan/work/outputs/london_streets_map.png", 
            dpi=150, bbox_inches="tight")
plt.show()
print("Map saved to outputs folder.")
```

```python
buildings = ox.features_from_place(
    "City of London, UK", 
    tags={"building": True}
)

buildings.to_file("/home/jovyan/work/data/london_buildings.gpkg", 
                  driver="GPKG")
print(f"Buildings saved: {len(buildings)}")
```

```python
pois = ox.features_from_place(
    "City of London, UK",
    tags={"amenity": True, "shop": True}
)

pois.to_file("/home/jovyan/work/data/london_pois.gpkg", 
             driver="GPKG")
print(f"POIs saved: {len(pois)}")
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

# Plot buildings in coral
buildings.plot(ax=ax, color="#D85A30", alpha=0.4, label="Buildings")

# Plot streets in blue
edges.plot(ax=ax, linewidth=0.8, color="#378ADD", alpha=0.9, label="Streets")

# Plot POIs as green dots
pois.plot(ax=ax, color="#1D9E75", markersize=3, alpha=0.7, label="POI")

ax.set_title("City of London — Streets, Buildings and POI", fontsize=14)
ax.set_axis_off()

plt.savefig("/home/jovyan/work/outputs/london_all_data.png",
            dpi=150, bbox_inches="tight")
plt.show()
print("Map saved.")
```
